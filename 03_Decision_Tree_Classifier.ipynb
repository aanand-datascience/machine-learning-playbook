**Post Prunning in Decision Tree**

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# useing small dataset because performing post prunning
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
```

```python
iris
```

```python
# input data
iris.data
```

```python
# output data
iris.target
```

```python
import seaborn as sns
```

```python
df=sns.load_dataset('iris')
```

```python
df.head()
```

```python
# independent and dependent faetures
x=df.iloc[:,:-1]
y=iris.target
```

```python
x
```

```python
y
```

```python
## train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.33,random_state=42)
```

```python
x_train
```

```python
y_train
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
# post prunning
# yaha pe parameter me kux nhi de rhe hai to bydefault Gini use ho rha hai
treemodel=DecisionTreeClassifier(max_depth=2)
```

```python
treemodel.fit(x_train,y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(20,20))
tree.plot_tree(treemodel,filled=True)
```

```python
#prediction
y_pred=treemodel.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
score=accuracy_score(y_pred,y_test)
```

```python
print(score)
```

```python
print(classification_report(y_pred,y_test))
```

**Pre-Prunning in Decision Tree**

```python
# Pre-prunning
parameter={
    'criterion':['gini', 'entropy', 'log_loss'], # default=”gini”
    'splitter':['best', 'random'],# default=”best”
    'max_depth': [1,2,3,4,5],
    'max_features':['auto','sqrt','log2']

}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
treemodel=DecisionTreeClassifier()
```

```python
cv=GridSearchCV(treemodel,parameter,cv=5,scoring='accuracy')
```

```python
cv.fit(x_train,y_train)
```

```python
cv.best_params_
```

```python
y_pred=cv.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
accuracy_score(y_pred,y_test)
```

```python
print(classification_report(y_pred,y_test))
```
