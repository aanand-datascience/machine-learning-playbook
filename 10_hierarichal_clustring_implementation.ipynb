**Ward:-** method ka use decide karne ke liye hota hai ki kaun-se do clusters ko merge karna chahiye.


Hierarchical Clustering mein Ward method woh do clusters merge karta hai jinke merge hone se within-cluster variance (data points ka apne cluster ke mean se spread) sabse kam increase hota hai.


Tumhare dendrogram se samjho
Maan lo 3 clusters hain:
- Cluster A
- Cluster B
- Cluster C


Ward method check karega ki A aur B ko merge karne se variance kitna badhega, aur A aur C ko merge karne se kitna badhega.

**uclidean Distance:-** batata hai ki do data points ek-dusre se kitni door hain.

Tumhare Hierarchical Clustering ke graph mein:


- Y-axis par distance kam hai: Dono points ya clusters paas hain.
- Y-axis par distance zyada hai: Dono points ya clusters door hain.

```python
## Agglomerative Clustring
## To construct a dendogram
import scipy.cluster.hierarchy as sc
## plot the dendogram
plt.figure(figsize=(20,7))
plt.title('Dendograms')


## Create dendograms
sc.dendrogram(sc.linkage(pca_x_train,method='ward'))
plt.xlabel('sample Index')
plt.ylabel('Eucledian Distance')
```

```python
from sklearn.cluster import AgglomerativeClustering
cluster=AgglomerativeClustering(n_clusters=2,metric='euclidean',linkage='ward')
cluster.fit(pca_x_train)
```

```python
cluster.labels_
```

```python
plt.scatter(pca_x_train[:,0],pca_x_train[:,1],c=cluster.labels_)
```

```python
## Performance matrics
## silhoutte score
from sklearn.metrics import silhouette_score
```

```python
silhouette_cofficients=[]

# Notice you start at 2 clusters for silhouette coefficient
for k in range(2,11):
  agglo=AgglomerativeClustering(n_clusters=k,metric='euclidean',linkage='ward')
  agglo.fit(pca_x_train)
  score=silhouette_score(pca_x_train,agglo.labels_)
  silhouette_cofficients.append(score)
```

```python
  silhouette_cofficients
```

```python
## plotting silhouette score
plt.plot(range(2,11),silhouette_cofficients)
# Haan, bilkul, x-axis pe sirf whole numbers (2, 3, 4, ... 10) dikhenge, koi decimal jaise 2.5 ya 3.5 nahi aayega.use of plt.xticks()
plt.xticks(range(2,11))
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Coefficients')
plt.show()
```
